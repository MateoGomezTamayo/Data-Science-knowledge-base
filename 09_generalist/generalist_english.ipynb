## 1 · Grammar — Subject-Verb Agreement <a id='grammar'></a>

### Q1: 'Maria and Juan _____ to the park every Saturday.'
> ✅ **`go`**

- **Plural subject** (Maria AND Juan) → base form verb (no -s)
- "every Saturday" → **simple present** (habitual action)

| Option | Why wrong |
|---|---|
| `go` ✅ | Plural subject → base form |
| `going` | ❌ Needs auxiliary: "are going" |
| `goes` | ❌ Third person **singular** only (he/she/it) |
| `went` | ❌ Past tense — contradicts "every Saturday" |

---

### Q2: 'The flowers _____ blooming beautifully in the garden.'
> ✅ **`are`**

- "The flowers" → **plural subject** → plural verb
- "blooming" → present participle → needs **present continuous** auxiliary
- No past context → present tense correct

| Option | Why wrong |
|---|---|
| `are` ✅ | Present tense plural |
| `were` | ❌ Past tense plural |
| `is` | ❌ Singular |
| `was` | ❌ Past tense singular |

---

### Q3: Present unreal conditional — which is correct?
> ✅ **`If she were there, she would see it happen.`**

**Structure:** `If + past subjunctive (were), would + base verb`

| Option | Type | Problem |
|---|---|---|
| *If she is there, she will see it* | Type 1 (real) | ❌ Not unreal |
| *If she was there, she would have seen it* | Mixed/Type 3 | ❌ Past unreal, not present |
| **If she were there, she would see it** ✅ | Type 2 (present unreal) | ✅ |
| *If she is there, she would see it* | Mixed | ❌ Mixes Type 1 + Type 2 |

> **Key rule:** Use **`were`** (not `was`) for ALL subjects in present unreal conditionals.

## 2 · Figurative Language <a id='figurative'></a>

### Figures of speech — quick reference

| Device | Definition | Example |
|---|---|---|
| **Metaphor** | Direct comparison (X IS Y) | *The air crackled with electricity* |
| **Simile** | Comparison using "like" or "as" | *He spoke like a seasoned lecturer* |
| **Personification** | Human qualities given to non-human | *The leaves danced and whispered* |
| **Oxymoron** | Two contradictory words fused together | *deafening silence* |
| **Epizeuxis** | Immediate repetition of a word for emphasis | *deep, deep* |
| **Anadiplosis** | Last word of clause = first word of next clause | *Fear→anger, anger→hate, hate→suffering* |
| **Anaphora** | Same word at the start of successive clauses | *We shall fight... we shall fight...* |
| **Juxtaposition** | Contrasting elements placed side by side | *vibrant market + lone beggar* |
| **Metafiction** | Narrative that self-consciously references its own fictional nature | Character speaks to the author |

---

### Q4: Metaphor for tension
> ✅ **`The air crackled with electricity as the meeting started.`**
- Metaphor (not simile — no "like/as")
- Describes tension without literal electricity

---

### Q5: Oxymoron for contradictory concepts
> ✅ **`The deafening silence was unnerving.`**
- "Deafening" (extremely loud) + "silence" (no sound) = contradiction in one phrase

---

### Q6: Epizeuxis for emphasis
> ✅ **`The silence was deep, deep, and all-encompassing.`**
- "deep, deep" = same word immediately repeated

---

### Q7: Anadiplosis
> ✅ **`Fear leads to anger, anger leads to hate, hate leads to suffering.`**
```
Fear → ANGER
       ANGER → HATE
               HATE → suffering
```

---

### Q8: Personification + visual imagery
> ✅ **`The leaves danced in the wind, whispering secrets to each other.`**
- "danced" → human physical action
- "whispering secrets" → human communication
- Both attributes enhance visual imagery

## 3 · Critical Reasoning <a id='critical'></a>

### Q9: Identify unrealistic claims — energy drink advertisement
> *"Our energy drink will give you endless energy and make you the most productive version of yourself without any need for sleep!"*

**Two unrealistic claims:**

**1. "Endless energy"** — No drink provides endless energy. Caffeine gives a temporary boost (1-3 hrs) followed by a crash. The human body has biological energy limits requiring rest and food.

**2. "Without any need for sleep"** — Physiologically impossible. Adults need 7-9 hours of sleep (CDC). No substance eliminates the biological need for sleep — caffeine only masks tiredness while sleep debt accumulates.

Both use absolute language ("endless", "without any need") that is unverifiable and misleading.

### Q10: Identify contradiction — plant-based diet passage
> *"Adopting a plant-based diet is beneficial for health as it reduces the risk of chronic diseases. However... a plant-based diet cannot support long-term health without supplements, making it less healthy than diets that include meat."*

**The contradiction:**
- Sentence 1: plant-based diet is **beneficial** (reduces chronic disease risk)
- Sentence 2: plant-based diet is **less healthy** than meat-inclusive diets

A diet cannot simultaneously reduce chronic disease AND be less healthy than alternatives.

**Additional issue:** "cannot" is an absolute claim — well-planned plant-based diets can meet all nutritional needs.

**Fix:** *"While a plant-based diet reduces chronic disease risk, it requires careful planning to avoid deficiencies. With appropriate supplementation, it can support long-term health comparably to balanced omnivorous diets."*

## 4 · Complex Passage Analysis <a id='analysis'></a>

### Framework for evaluating a passage

When asked to evaluate a passage's logical consistency and depth, check for:

| Check | Question to ask |
|---|---|
| **Unsupported claims** | Is the assertion backed by evidence or examples? |
| **Absolute language** | Words like "all", "never", "universal", "cannot" — are they justified? |
| **Missing dimensions** | Are economic, social, ethical, or technical aspects ignored? |
| **Internal contradictions** | Does a later claim undermine an earlier one? |
| **Vague conclusions** | Does the ending restate the problem or provide actionable recommendations? |
| **Logical gaps** | Does the argument skip steps or assume unstated premises? |

---

### Q11: Renewable energy passage
**Issues identified:**
1. "Universal adoption" — unsupported, absolute claim
2. "Policies are being developed" — no specific examples cited
3. Economic trade-off mentioned but not explored
4. "Global efforts required" — ignores historical difficulty of coordination
5. Missing: social equity dimension (job losses, energy poverty)
6. Conclusion is circular — restates problem instead of proposing solutions

---

### Q12: Telemedicine passage
**Issues identified:**
1. "Universal adoption will transform" — no evidence provided
2. Inequality point raised but immediately dropped — no exploration of who is affected
3. Depersonalization claim doesn't distinguish between care types (routine vs. mental health)
4. "Security concerns remain unresolved" — ignores HIPAA, GDPR frameworks already in place
5. Conclusion "balanced integration" adds nothing specific

---

### Q13: Juxtaposition — marketplace and beggar
**The technique:** Placing "vibrant colors and cheerful voices" alongside "a lone beggar" in the same location.

**Effect on duality theme:**
- Poverty and prosperity coexist in the same space — cannot be compartmentalized
- Forces emotional confrontation rather than intellectual abstraction
- The market continues "undisturbed" → implies normalization of inequality

---

### Q14: Metafiction — pen hovering, character speaks
**The technique:** Author visible inside the scene; character directly addresses the author.

**Effect on creation theme:**
- Exposes creation as a continuous decision (pen paused = indeterminacy)
- Character claims co-authorship by intervening

**Effect on agency theme:**
- Character refuses passivity — leaning forward = exercising will
- Reader also becomes a co-creator: "a beginning you have yet to imagine"
- Challenges the author's absolute power over their creation

## 5 · Multiple Choice — Select the Best Option <a id='mc'></a>

**Pattern:** Always pick the option that:
- ✅ Answers ALL parts of the question
- ✅ Is balanced (acknowledges both benefits and concerns)
- ✅ Uses qualified language ("can", "may", "raises concerns") not absolutes
- ❌ Avoid options that say "no significant concerns" when the question asks about concerns

---

### Q15: Health benefit of fruits and vegetables
> ✅ *"Eating a diet rich in fruits and vegetables provides essential vitamins and minerals that boost overall health and reduce the risk of chronic diseases."*
- ❌ Wrong: "**guarantees** weight loss and **instantly** improves fitness" — absolute, unverifiable claims

---

### Q16: Autonomous decision-making in corporate management
> ✅ *"Can enhance efficiency and reduce human error... but raises ethical dilemmas such as accountability issues, programmed bias, and impacts on workforce morale."*
- ❌ Wrong: "No significant ethical implications" — ignores what the question explicitly asks about

---

### Q17: AI-driven surveillance — international law and regional stability
> ✅ *"Can provide strategic surveillance... however, they challenge existing maritime laws, raise concerns about territorial waters, and could provoke destabilizing actions."*
- ❌ Wrong: "Mainly a technical advancement with no significant legal or stability concerns" — dismisses the entire question

---

### Answer pattern summary

| ❌ Red flags in wrong answers | ✅ Green flags in correct answers |
|---|---|
| "No significant concerns" | "Can enhance... but raises..." |
| "Guarantees", "instantly", "always" | "May", "potentially", "can" |
| Only covers one dimension | Covers all dimensions asked |
| Dismisses the question's premise | Directly addresses what is asked |
