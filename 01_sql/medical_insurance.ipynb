# Medical Insurance Price Prediction
**Dataset:** [Kaggle — Medical Insurance Price Prediction](https://www.kaggle.com/datasets/harishkumardatalab/medical-insurance-price-prediction)

**Goal:** Predict individual medical insurance `charges` from demographic and lifestyle attributes.

| Column | Type | Description |
|--------|------|-------------|
| `age` | int | Age of the beneficiary |
| `sex` | str | Gender (male / female) |
| `bmi` | float | Body Mass Index |
| `children` | int | Number of dependents covered |
| `smoker` | str | Smoker status (yes / no) |
| `region` | str | US residential region (4 values) |
| `charges` | float | **Target** — medical costs billed |
