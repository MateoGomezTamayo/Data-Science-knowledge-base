## Analyzing the Automobile Sales data of XYZAutomotives

Historical automobile sales data used to study how sales, advertising expenditure, GDP, unemployment and vehicle type behave across recession and non-recession periods.

Libraries used: **Matplotlib, Seaborn and Pandas**.

## Import Required Libraries and Load the Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Load the historical automobile sales dataset
df = pd.read_csv(
    "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/"
    "IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/"
    "historical_automobile_sales.csv"
)

print("Data loaded:", df.shape)
print("Columns:", df.columns.tolist())
df.head()
```

## TASK 1.1 — Line chart: how automobile sales fluctuate from year to year

```python
# Average automobile sales per year
df_line = df.groupby(df["Year"])["Automobile_Sales"].mean()

plt.figure(figsize=(10, 6))
df_line.plot(kind="line")
plt.xlabel("Year")
plt.ylabel("Average Automobile Sales")
plt.title("Automobile Sales Fluctuation from Year to Year")
plt.show()
```

## TASK 1.2 — Line chart: advertising expenditure vs automobile sales during non-recession periods

```python
# Non-recession data only
df_non_rec = df[df["Recession"] == 0]

# Average automobile sales and average advertising expenditure by year
df_trends = df_non_rec.groupby("Year", as_index=False).agg(
    Avg_Sales=("Automobile_Sales", "mean"),
    Avg_Ad_Spend=("Advertising_Expenditure", "mean"),
)

plt.figure(figsize=(12, 6))
sns.lineplot(data=df_trends, x="Year", y="Avg_Sales", marker="o", color="green", label="Average Sales")
sns.lineplot(data=df_trends, x="Year", y="Avg_Ad_Spend", marker="s", linestyle="--", color="blue", label="Average Advertising Expenditure")
plt.xlabel("Year")
plt.ylabel("Average Value")
plt.title("Advertising Expenditure vs Automobile Sales over the Years (Non-Recession)")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

## TASK 1.3 — Bar chart: vehicle-wise sales trend during recession vs non-recession periods

```python
# Average sales by recession status and vehicle type
grouped_df = df.groupby(["Recession", "Vehicle_Type"])["Automobile_Sales"].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x="Recession", y="Automobile_Sales", hue="Vehicle_Type", data=grouped_df)
plt.xticks(ticks=[0, 1], labels=["Non-Recession", "Recession"])
plt.xlabel("Period")
plt.ylabel("Average Automobile Sales")
plt.title("Vehicle-Wise Sales during Recession and Non-Recession Periods")
plt.show()
```

## TASK 1.4 — Subplots: GDP variation during recession vs non-recession periods (line plots)

```python
rec_data = df[df["Recession"] == 1]
non_rec_data = df[df["Recession"] == 0]

fig = plt.figure(figsize=(12, 6))
ax0 = fig.add_subplot(1, 2, 1)
ax1 = fig.add_subplot(1, 2, 2)

sns.lineplot(x="Year", y="GDP", data=rec_data, label="Recession", ax=ax0)
ax0.set_xlabel("Year")
ax0.set_ylabel("GDP")
ax0.set_title("GDP Variation during Recession Period")

sns.lineplot(x="Year", y="GDP", data=non_rec_data, label="Non-Recession", ax=ax1)
ax1.set_xlabel("Year")
ax1.set_ylabel("GDP")
ax1.set_title("GDP Variation during Non-Recession Period")

plt.tight_layout()
plt.show()
```

## TASK 1.5 — Bubble plot: impact of seasonality on automobile sales

```python
non_rec_data = df[df["Recession"] == 0]

size = non_rec_data["Seasonality_Weight"]  # bubble size

plt.figure(figsize=(12, 6))
sns.scatterplot(data=non_rec_data, x="Month", y="Automobile_Sales", size=size, hue=size, legend=False)
plt.xlabel("Month")
plt.ylabel("Automobile Sales")
plt.title("Seasonality Impact on Automobile Sales")
plt.show()
```

## TASK 1.6 — Scatter plot: average vehicle price vs sales volume during recessions

```python
rec_data = df[df["Recession"] == 1]

plt.figure(figsize=(10, 6))
plt.scatter(rec_data["Price"], rec_data["Automobile_Sales"], alpha=0.6)
plt.xlabel("Average Vehicle Price")
plt.ylabel("Automobile Sales")
plt.title("Relationship between Average Vehicle Price and Sales Volume during Recessions")
plt.show()
```

## TASK 1.7 — Pie chart: advertising expenditure during recession vs non-recession periods

```python
Rdata = df[df["Recession"] == 1]
NRdata = df[df["Recession"] == 0]

# Total advertising expenditure for each period
RAtotal = Rdata["Advertising_Expenditure"].sum()
NRAtotal = NRdata["Advertising_Expenditure"].sum()

plt.figure(figsize=(8, 6))
labels = ["Recession", "Non-Recession"]
sizes = [RAtotal, NRAtotal]
plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Advertising Expenditure during Recession and Non-Recession Periods")
plt.show()
```

## TASK 1.8 — Pie chart: total advertisement expenditure for each vehicle type during recession

```python
Rdata = df[df["Recession"] == 1]

# Advertising expenditure by vehicle type during recessions
VTexpenditure = Rdata.groupby("Vehicle_Type")["Advertising_Expenditure"].sum()

plt.figure(figsize=(8, 6))
plt.pie(VTexpenditure.values, labels=VTexpenditure.index, autopct="%1.1f%%", startangle=90)
plt.title("Total Advertisement Expenditure for Each Vehicle Type during Recession")
plt.show()
```

## TASK 1.9 — Line chart: effect of unemployment rate on vehicle type and sales during recession

```python
df_rec = df[df["Recession"] == 1]

plt.figure(figsize=(16, 6))
sns.lineplot(
    data=df_rec,
    x="unemployment_rate",
    y="Automobile_Sales",
    hue="Vehicle_Type",
    style="Vehicle_Type",
    markers=True,
)
plt.xlabel("Unemployment Rate")
plt.ylabel("Automobile Sales")
plt.title("Effect of Unemployment Rate on Vehicle Type and Sales during Recession")
plt.legend(title="Vehicle Type")
plt.tight_layout()
plt.show()
```
