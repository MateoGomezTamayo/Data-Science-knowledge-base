## 1 · Operators — `^` vs `**` <a id='operators'></a>

### Bug
```python
squares.append(num^2)   # ❌ ^ is bitwise XOR, not exponentiation
```

### Fix
```python
squares.append(num**2)  # ✅ ** is the exponentiation operator
```

### Operator table

| Operator | Name | Example | Result |
|---|---|---|---|
| `**` | Exponentiation | `3 ** 2` | `9` |
| `^` | Bitwise XOR | `3 ^ 2` | `1` |
| `*` | Multiplication | `3 * 2` | `6` |
| `//` | Floor division | `7 // 2` | `3` |
| `%` | Modulo | `7 % 2` | `1` |

> **Why `3 ^ 2 = 1`?**  
> XOR compares bits: `011 XOR 010 = 001` → `1`

````python
# ── Bug: ^ is XOR, not power ────────────────────────────────────
numbers = [1, 2, 3, 4]

# ❌ Wrong — uses bitwise XOR
wrong = []
for num in numbers:
    wrong.append(num ^ 2)     # XOR with 2, not squared
print('❌ With ^  (XOR):', wrong)

# ✅ Correct — exponentiation
squares = []
for num in numbers:
    squares.append(num ** 2)
print('✅ With ** (power):', squares)

# ✅ Pythonic version
squares_lc = [num ** 2 for num in numbers]
print('✅ List comprehension:', squares_lc)

# Proof: XOR vs power on 3
print(f'\n3 ^ 2  = {3 ^ 2}   ← XOR (01 1 XOR 01 0 = 00 1)')
print(f'3 ** 2 = {3 ** 2}   ← exponentiation')
````

## 2 · String Indexing in Loops <a id='string-index'></a>

### Question
```python
words = ['apple', 'banana', 'cherry']
result = ''
for word in words:
    result += word[0]
print(result)
```

### Output: `abc`

**Trace:**

| Iteration | `word` | `word[0]` | `result` |
|---|---|---|---|
| 1 | `'apple'` | `'a'` | `'a'` |
| 2 | `'banana'` | `'b'` | `'ab'` |
| 3 | `'cherry'` | `'c'` | `'abc'` |

`word[0]` accesses the **first character** (index 0) of each string.  
`result += char` is string concatenation — equivalent to `result = result + char`.

````python
# ── String indexing demo ────────────────────────────────────────
words = ['apple', 'banana', 'cherry']
result = ''

for word in words:
    print(f'  word={word!r:10s}  word[0]={word[0]!r}  result so far: {result + word[0]!r}')
    result += word[0]

print(f'\nFinal output: {result!r}')

# ── String index reference ─────────────────────────────────────
s = 'hello'
print(f'\nString indexing on {s!r}:')
print(f'  s[0]  = {s[0]!r}   ← first character')
print(f'  s[-1] = {s[-1]!r}   ← last character')
print(f'  s[1:3]= {s[1:3]!r}  ← slice [start:end]')

# ── Pythonic equivalent ────────────────────────────────────────
initials = ''.join(word[0] for word in words)
print(f'\nPythonic (join + generator): {initials!r}')
````

## 3 · List Methods — `.append()` vs `.add()` <a id='list-methods'></a>

### Bug
```python
doubled_numbers.add(n * 2)    # ❌ lists have no .add() method
```

### Fix
```python
doubled_numbers.append(n * 2) # ✅ lists use .append()
```

### Method comparison

| Method | Works on | Purpose |
|---|---|---|
| `.append(x)` | `list` ✅ | Add one element to the end |
| `.extend([x,y])` | `list` | Add multiple elements |
| `.insert(i, x)` | `list` | Insert at position i |
| `.add(x)` | `set` | Add element to a set |
| `.update([x,y])` | `set` | Add multiple elements to a set |

````python
# ── .append() vs .add() ─────────────────────────────────────────
numbers = [1, 2, 3, 4, 5]

# ❌ Wrong — lists have no .add()
try:
    bad_list = []
    bad_list.add(10)
except AttributeError as e:
    print(f'❌ Error on list.add(): {e}')

# ✅ Correct — use .append() for lists
doubled_numbers = []
for n in numbers:
    doubled_numbers.append(n * 2)
print(f'\n✅ doubled_numbers: {doubled_numbers}')

# ✅ .add() works on SETS
my_set = {1, 2, 3}
my_set.add(4)
print(f'✅ set.add() works:  {my_set}')

# ── All list add methods ───────────────────────────────────────
lst = [1, 2, 3]
lst.append(4)          # → [1,2,3,4]
lst.extend([5, 6])     # → [1,2,3,4,5,6]
lst.insert(0, 0)       # → [0,1,2,3,4,5,6]
print(f'\nAfter append+extend+insert: {lst}')

# ✅ Pythonic
doubled = [n * 2 for n in numbers]
print(f'List comprehension: {doubled}')
````

## 4 · if / elif / else — First Match Wins <a id='if-elif'></a>

### Question
```python
x = 10
if x > 5:
    print('Greater than 5')
elif x > 3:
    print('Greater than 3')
else:
    print('3 or less')
```

### Output: `Greater than 5`

**Rule: Python evaluates top-to-bottom and stops at the FIRST True condition.**

```
if x > 5:    ← 10 > 5 = True ✅ → executes, skips everything below
elif x > 3:  ← never evaluated
else:        ← never reached
```

Even though `10 > 3` is also `True`, `elif` only runs if **all previous conditions were `False`**.

````python
# ── if/elif/else: first match wins ─────────────────────────────
def classify(x):
    if x > 5:
        return 'Greater than 5'
    elif x > 3:
        return 'Greater than 3'
    else:
        return '3 or less'

test_values = [10, 4, 3, 0, -1]
for v in test_values:
    print(f'  x={v:3d}  →  {classify(v)}')

# ── Visualise which branch is taken ───────────────────────────
print('\n── Detailed trace for x=10 ──')
x = 10
print(f'  x > 5  : {x} > 5  = {x > 5}  ← first condition True → executes')
print(f'  x > 3  : never evaluated (first branch already matched)')

print('\n── Detailed trace for x=4 ──')
x = 4
print(f'  x > 5  : {x} > 5  = {x > 5}  ← False → check next')
print(f'  x > 3  : {x} > 3  = {x > 3}  ← True  → executes')

print('\n── Detailed trace for x=2 ──')
x = 2
print(f'  x > 5  : {x} > 5  = {x > 5}  ← False')
print(f'  x > 3  : {x} > 3  = {x > 3}  ← False')
print(f'  else   : all conditions False → else executes')
````

## 5 · Common Python Bugs — Quick Reference <a id='bugs'></a>

| Bug | Wrong | Correct | Error type |
|---|---|---|---|
| Exponentiation | `num^2` | `num**2` | Wrong operator (XOR ≠ power) |
| Add to list | `lst.add(x)` | `lst.append(x)` | `AttributeError` |
| Off-by-one | `range(1, n)` | `range(1, n+1)` | Missing last element |
| Integer division | `5 / 2 → 2.5` | `5 // 2 → 2` | Use `//` for floor division |
| String comparison | `x = = y` | `x == y` | `SyntaxError` |
| Mutable default arg | `def f(lst=[])` | `def f(lst=None)` | Shared state bug |
| `=` vs `==` | `if x = 5` | `if x == 5` | Assignment in condition |

````python
# ── All bugs demonstrated and fixed ────────────────────────────

# 1. ^ is XOR, ** is power
print('1. Exponentiation:')
print(f'   3^2  = {3^2}  (XOR)  |  3**2 = {3**2}  (power)')

# 2. list.append vs set.add
print('\n2. Adding to collections:')
lst = [1, 2]; lst.append(3);   print(f'   list.append: {lst}')
st  = {1, 2}; st.add(3);       print(f'   set.add:     {st}')

# 3. Off-by-one
print('\n3. Off-by-one in range:')
print(f'   range(1,5)  = {list(range(1,5))}    ← stops before 5')
print(f'   range(1,6)  = {list(range(1,6))}  ← includes 5')

# 4. Division types
print('\n4. Division:')
print(f'   5 / 2  = {5/2}   (float division)')
print(f'   5 // 2 = {5//2}     (floor division)')
print(f'   5 % 2  = {5%2}     (modulo / remainder)')

# 5. Mutable default argument trap
print('\n5. Mutable default arg bug:')
def bad_append(item, lst=[]):   # shared across all calls!
    lst.append(item)
    return lst

def good_append(item, lst=None):  # safe
    if lst is None: lst = []
    lst.append(item)
    return lst

print(f'   bad call 1: {bad_append(1)}')
print(f'   bad call 2: {bad_append(2)}  ← still has [1]!')
print(f'   good call1: {good_append(1)}')
print(f'   good call2: {good_append(2)}  ← fresh list ✅')
````

## 6 · pandas — Finding Max/Min Values & Their Labels <a id='idxmax'></a>

### Question
> **Which food item has maximum sodium content?**

### 3-step workflow

```python
# Step 1 — Explore: find the max value
df['Sodium'].describe()
# max    3600.0

# Step 2 — Get the INDEX (row label) of the max
df['Sodium'].idxmax()
# → 82

# Step 3 — Look up the item name at that row
df.at[82, 'Item']
# → 'Chicken McNuggets (40 piece)'
```

### One-liner

```python
df.loc[df['Sodium'].idxmax(), 'Item']
# → 'Chicken McNuggets (40 piece)'
```

### Key methods

| Method | Returns | Example |
|---|---|---|
| `.describe()` | Summary stats (min, max, mean, std, %tiles) | `df['Sodium'].describe()` |
| `.idxmax()` | **Row index** of the maximum value | `df['Sodium'].idxmax()` → `82` |
| `.idxmin()` | Row index of the minimum value | `df['Sodium'].idxmin()` |
| `.max()` | The maximum value itself | `df['Sodium'].max()` → `3600.0` |
| `.at[row, col]` | Single value by row + column label | `df.at[82, 'Item']` |
| `.loc[mask, col]` | Filter rows by condition | `df.loc[df.Sodium==df.Sodium.max(), 'Item']` |

````python
# ── Finding max item in pandas ─────────────────────────────────
import pandas as pd
import numpy as np

# Simulate a food nutrition dataset
rng = np.random.default_rng(42)
df = pd.DataFrame({
    'Item':    ['Burger', 'Salad', 'Pizza', 'Fries', 'Chicken McNuggets (40 piece)',
                'Wrap', 'Soda', 'Shake', 'Hotdog', 'Soup'],
    'Sodium':  [1000, 300, 950, 400, 3600, 750, 50, 180, 860, 620],
    'Calories':[540,  150, 800, 320, 1770, 410, 140, 530, 290, 200],
})

# ── Step 1: describe() ────────────────────────────────────────
print('── Step 1: df[\"Sodium\"].describe() ──')
print(df['Sodium'].describe())

# ── Step 2: idxmax() → row index ─────────────────────────────
idx = df['Sodium'].idxmax()
print(f'\n── Step 2: idxmax() = {idx}')

# ── Step 3: look up the name ──────────────────────────────────
item = df.at[idx, 'Item']
sodium = df.at[idx, 'Sodium']
print(f'\n── Step 3: df.at[{idx}, \"Item\"] = \"{item}\"')
print(f'           Sodium content: {sodium} mg')

# ── One-liner ─────────────────────────────────────────────────
print('\n── One-liner ──')
print(df.loc[df['Sodium'].idxmax(), 'Item'])

# ── idxmax vs idxmin ─────────────────────────────────────────
print('\n── Max and Min sodium items ──')
print(f'  Most sodium : {df.loc[df["Sodium"].idxmax(), "Item"]}  ({df["Sodium"].max()} mg)')
print(f'  Least sodium: {df.loc[df["Sodium"].idxmin(), "Item"]}  ({df["Sodium"].min()} mg)')

# ── Top 3 highest sodium ─────────────────────────────────────
print('\n── Top 3 highest sodium ──')
display(df.nlargest(3, 'Sodium')[['Item', 'Sodium']].reset_index(drop=True))
````
