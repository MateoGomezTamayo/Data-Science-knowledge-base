## CHI SQUARE 
The chi-square test is a statistical method used to determine if there is a significant association between two categorical variables. This test is widely used in various fields, including social sciences, marketing, and healthcare, to analyze survey data, experimental results, and observational studies.

```python
import pandas as pd
from scipy.stats import chi2_contingency
# Create the contingency table
data = [[20, 30],  # Male: [Like, Dislike]
        [25, 25]]  # Female: [Like, Dislike]
# Create a DataFrame for clarity
df = pd.DataFrame(data, columns=["Like", "Dislike"], index=["Male", "Female"])
# Perform the Chi-Square Test
chi2, p, dof, expected = chi2_contingency(df)
# Display results
print("Chi-square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("P-value:", p)
print("Expected Frequencies:\n", expected) 
```

```python
import numpy as np

import matplotlib.pyplot as plt

# Example: Model comparison with accuracy vs efficiency trade-off
models = ['Model A', 'Model B', 'Model C', 'Model D']
accuracy = [0.92, 0.94, 0.89, 0.91]
inference_time_ms = [50, 500, 20, 150]  # milliseconds
vram_gb = [2, 5, 1.5, 3]
cost_per_1k_inferences = [10, 50, 5, 20]

# Create figure with subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Accuracy vs Inference Time (Pareto frontier visualization)
ax1 = axes[0, 0]
scatter = ax1.scatter(inference_time_ms, accuracy, s=300, alpha=0.6, c=range(len(models)), cmap='viridis')
for i, model in enumerate(models):
    ax1.annotate(model, (inference_time_ms[i], accuracy[i]), xytext=(5, 5), textcoords='offset points')
ax1.set_xlabel('Inference Time (ms)', fontsize=11, fontweight='bold')
ax1.set_ylabel('Accuracy', fontsize=11, fontweight='bold')
ax1.set_title('Accuracy vs Speed Trade-off', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3)

# 2. Accuracy vs VRAM Usage
ax2 = axes[0, 1]
ax2.scatter(vram_gb, accuracy, s=300, alpha=0.6, c=range(len(models)), cmap='viridis')
for i, model in enumerate(models):
    ax2.annotate(model, (vram_gb[i], accuracy[i]), xytext=(5, 5), textcoords='offset points')
ax2.set_xlabel('VRAM Required (GB)', fontsize=11, fontweight='bold')
ax2.set_ylabel('Accuracy', fontsize=11, fontweight='bold')
ax2.set_title('Accuracy vs Memory Trade-off', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)

# 3. Speed vs Cost (bubble chart with size = accuracy)
ax3 = axes[1, 0]
sizes = [acc * 1000 for acc in accuracy]  # Scale for visibility
ax3.scatter(inference_time_ms, cost_per_1k_inferences, s=sizes, alpha=0.6, c=range(len(models)), cmap='viridis')
for i, model in enumerate(models):
    ax3.annotate(model, (inference_time_ms[i], cost_per_1k_inferences[i]), xytext=(5, 5), textcoords='offset points')
ax3.set_xlabel('Inference Time (ms)', fontsize=11, fontweight='bold')
ax3.set_ylabel('Cost per 1K Inferences ($)', fontsize=11, fontweight='bold')
ax3.set_title('Speed vs Cost (bubble size = accuracy)', fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3)

# 4. Radar/Multi-dimensional trade-off (normalized comparison)
ax4 = axes[1, 1]
ax4.axis('off')

# Normalize metrics to 0-1 scale (inverted where lower is better)
norm_accuracy = np.array(accuracy) / max(accuracy)
norm_speed = 1 - np.array(inference_time_ms) / max(inference_time_ms)
norm_memory = 1 - np.array(vram_gb) / max(vram_gb)
norm_cost = 1 - np.array(cost_per_1k_inferences) / max(cost_per_1k_inferences)

# Create table
table_data = []
table_data.append(['Model', 'Accuracy', 'Speed', 'Memory', 'Cost'])
for i, model in enumerate(models):
    table_data.append([
        model,
        f'{norm_accuracy[i]:.2f}',
        f'{norm_speed[i]:.2f}',
        f'{norm_memory[i]:.2f}',
        f'{norm_cost[i]:.2f}'
    ])

table = ax4.table(cellText=table_data, cellLoc='center', loc='center',
                  colWidths=[0.15, 0.15, 0.15, 0.15, 0.15])
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1, 2)

# Color header row
for i in range(5):
    table[(0, i)].set_facecolor('#4CAF50')
    table[(0, i)].set_text_props(weight='bold', color='white')

# Color cells based on values (normalized scores)
for i in range(1, len(models) + 1):
    for j in range(1, 5):
        val = float(table_data[i][j])
        color_intensity = val
        table[(i, j)].set_facecolor(plt.cm.RdYlGn(color_intensity))

ax4.set_title('Normalized Trade-off Scores (0-1)', fontsize=12, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
```

```python
from tqdm import tqdm
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import PolynomialFeatures
```

```python
try:
    # Available when the notebook runs in Pyodide/JupyterLite
    from pyodide.http import pyfetch

except ModuleNotFoundError:
    # Fallback for standard Jupyter running with CPython
    import urllib.request

    class _Response:
        def __init__(self, response):
            self.status = response.status
            self._content = response.read()

        async def bytes(self):
            return self._content

    async def pyfetch(url):
        response = urllib.request.urlopen(url)
        return _Response(response)

async def download(url, filename):
    response = await pyfetch(url)
    if response.status == 200:
        with open(filename, "wb") as f:
            f.write(await response.bytes())
```

```python
filepath = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/laptop_pricing_dataset_mod2.csv'
```

```python
await download(filepath, "laptops.csv")
file_name="laptops.csv"
```

```python
df = pd.read_csv(file_name, header=0)
```

```python
df.drop(['Unnamed: 0', 'Unnamed: 0.1'], axis=1, inplace=True)
```

```python
y_data = df['Price']
x_data = df.drop('Price',axis=1)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.10, random_state=1)
print("number of test samples :", x_test.shape[0])
print("number of training samples:",x_train.shape[0])
```

```python
lre=LinearRegression()
lre.fit(x_train[['CPU_frequency']], y_train)
print(lre.score(x_test[['CPU_frequency']], y_test))
print(lre.score(x_train[['CPU_frequency']], y_train))
```

```python
import builtins

Rcross = cross_val_score(lre, x_train[['CPU_frequency']], y_train, cv=4)
builtins.print(
    "The mean of the folds are",
    Rcross.mean(),
    "and the standard deviation is",
    Rcross.std()
)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.5, random_state=0)
# fixing random_state to a fixed quantity helps maintain uniformity between multiple 
# executions of the code.
```

```python
lre = LinearRegression()
Rsqu_test = []
order = [1, 2, 3, 4, 5]
for n in order:
    pr = PolynomialFeatures(degree=n)
    x_train_pr = pr.fit_transform(x_train[['CPU_frequency']])
    x_test_pr = pr.fit_transform(x_test[['CPU_frequency']])    
    lre.fit(x_train_pr, y_train)
    Rsqu_test.append(lre.score(x_test_pr, y_test))
```

```python
plt.plot(order, Rsqu_test)
plt.xlabel('order')
plt.ylabel('R^2')
plt.title('R^2 Using Test Data')
```

```python
pr=PolynomialFeatures(degree=2)
x_train_pr=pr.fit_transform(x_train[['CPU_frequency', 'RAM_GB', 'Storage_GB_SSD', 'CPU_core', 'OS', 'GPU', 'Category']])
x_test_pr=pr.fit_transform(x_test[['CPU_frequency', 'RAM_GB', 'Storage_GB_SSD', 'CPU_core', 'OS', 'GPU', 'Category']])
```

```python
Rsqu_test = []
Rsqu_train = []
Alpha = np.arange(0.001,1,0.001)
pbar = tqdm(Alpha)

for alpha in pbar:
    RigeModel = Ridge(alpha=alpha) 
    RigeModel.fit(x_train_pr, y_train)
    test_score, train_score = RigeModel.score(x_test_pr, y_test), RigeModel.score(x_train_pr, y_train)
    pbar.set_postfix({"Test Score": test_score, "Train Score": train_score})
    Rsqu_test.append(test_score)
    Rsqu_train.append(train_score)
```

```python
plt.figure(figsize=(10, 6))  
plt.plot(Alpha, Rsqu_test, label='validation data')
plt.plot(Alpha, Rsqu_train, 'r', label='training Data')
plt.xlabel('alpha')
plt.ylabel('R^2')
plt.ylim(0, 1)
plt.legend()
```

```python
parameters1= [{'alpha': [0.0001,0.001,0.01, 0.1, 1, 10]}]
```

```python
RR=Ridge()
Grid1 = GridSearchCV(RR, parameters1,cv=4)
```

```python
Grid1.fit(x_train[['CPU_frequency', 'RAM_GB', 'Storage_GB_SSD', 'CPU_core', 'OS', 'GPU', 'Category']], y_train)
```

```python
BestRR=Grid1.best_estimator_
print(BestRR.score(x_test[['CPU_frequency', 'RAM_GB', 'Storage_GB_SSD', 'CPU_core','OS','GPU','Category']], y_test))
```
